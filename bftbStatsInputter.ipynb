```python
allPlayers = []
for (event, teams) in eventPlayersDict.items():
    for team, players in teams.items():
        for player in players:
            if player[0] not in allPlayers:
                allPlayers.append(player[0])

print(allPlayers)
```

```python
def get_online_uuid(username: str) -> str | None:
    url = f"https://api.minecraftservices.com/minecraft/profile/lookup/name/{username}"
    response = requests.get(url)
    
    if response.status_code == 200:
        raw_id = response.json()["id"]
        # convert the 32-char hex string into a standard dashed UUID
        raw_id = f"{raw_id[:8]}-{raw_id[8:12]}-{raw_id[12:16]}-{raw_id[16:20]}-{raw_id[20:]}"

        return raw_id
    return None
```

```python
playerUUIDS = {}
for player in allPlayers:
    uuid = get_online_uuid(player)
    if uuid:
        playerUUIDS[player] = uuid
    else:
        print(f"Could not retrieve UUID for player: {player}")
```

```python
# put data to event_players.csv with format event,uuid,team

with open("event_players.csv", "w", newline="") as f:
    toWrite = [["event", "uuid", "team"]]


    for event, teams in eventPlayersDict.items():
        for team, players in teams.items():
            for player in players:
                if player[0] in playerUUIDS:
                    toWrite.append([event, playerUUIDS[player[0]], team])
                else:
                    print(f"Skipping player {player} for event {event} and team {team} due to missing UUID.")

    print(toWrite)
    writer = csv.writer(f)
    writer.writerows(toWrite)
```

```python
# input all scores to event_scores.csv with format event,gamenum,uuid,score

with open("event_scores.csv", "w", newline="") as f:
    toWriteScores = [["event", "gamenum", "uuid", "score"]]

    for event, teams in eventPlayersDict.items():
        for team, players in teams.items():
            for player in players:
                if player[0] in playerUUIDS:
                    for game_num in range(1, len(player)):
                        score = player[game_num]
                        toWriteScores.append([event, str(game_num), playerUUIDS[player[0]], round(float(score), 2)])
                else:
                    print(f"Skipping player {player} for event {event} and team {team} due to missing UUID.")

    print(toWriteScores)
    writer = csv.writer(f)
    writer.writerows(toWriteScores)
```
